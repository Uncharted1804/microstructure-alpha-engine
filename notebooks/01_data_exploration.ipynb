Cell 1 — Imports:

```python
import sys
sys.path.insert(0, '..')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from src.data_collection.lob_reconstructor import (
    load_depth_data, load_trade_data, add_derived_columns
)

plt.style.use('dark_background')
%matplotlib inline

RUN_ID   = "20260524_170957"   # ← change this
DATA_DIR = Path("../data")
```

Cell 2 — Load data:


```python
depth  = load_depth_data(RUN_ID, DATA_DIR)
trades = load_trade_data(RUN_ID, DATA_DIR)
depth  = add_derived_columns(depth)

print(f"Depth snapshots : {len(depth):,}")
print(f"Trades          : {len(trades):,}")
print(f"Duration        : {(depth['timestamp_ms'].iloc[-1] - depth['timestamp_ms'].iloc[0])/60000:.1f} minutes")
print(f"Mid-price range : {depth['mid_price'].min():.2f} – {depth['mid_price'].max():.2f}")
print(f"Avg spread      : {depth['spread'].mean():.4f} ({depth['relative_spread_bps'].mean():.2f} bps)")
```

Cell 3 — Price and spread chart:

```python
fig, axes = plt.subplots(3, 1, figsize=(14, 9), sharex=True)

axes[0].plot(depth['datetime'], depth['mid_price'], lw=0.8, color='#00ff88')
axes[0].set_ylabel('Mid Price')
axes[0].set_title('BTC/USDT Mid-Price')

axes[1].plot(depth['datetime'], depth['relative_spread_bps'], lw=0.5, color='#ff6b6b', alpha=0.8)
axes[1].set_ylabel('Spread (bps)')
axes[1].set_title('Bid-Ask Spread')

axes[2].plot(depth['datetime'], depth['queue_imbalance'], lw=0.5, color='#ffa500', alpha=0.8)
axes[2].axhline(0, color='white', lw=0.5, ls='--')
axes[2].set_ylabel('Queue Imbalance')
axes[2].set_title('Queue Imbalance (+1 = all bids, -1 = all asks)')

plt.tight_layout()
plt.savefig('../data/processed/01_data_overview.png', dpi=150, bbox_inches='tight')
plt.show()
```
