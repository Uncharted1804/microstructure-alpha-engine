```python
# Cell 1
import sys; sys.path.insert(0, '..')
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from src.data_collection.lob_reconstructor import load_trade_data
from src.hawkes.hawkes_calibration import (
    extract_trade_timestamps, calibrate_hawkes_mle_manual,
    compute_hawkes_intensity, fit_rolling_hawkes
)

plt.style.use('dark_background')
%matplotlib inline

RUN_ID = "YOUR_RUN_ID_HERE"
trades = load_trade_data(RUN_ID, Path("../data"))
```

```python
# Cell 2 — Calibrate (cap at 3000 trades for speed)
ts = extract_trade_timestamps(trades)[:3000]
print(f"Calibrating on {len(ts)} trades...")
params = calibrate_hawkes_mle_manual(ts)
```

```python
# Cell 3 — Plot intensity for first 60 seconds
sample_ts = ts[ts <= 60]
eval_t, intensity = compute_hawkes_intensity(sample_ts, params)

fig, axes = plt.subplots(2, 1, figsize=(13, 6), sharex=True)
axes[0].plot(eval_t, intensity, color='#00bfff', lw=1.0)
axes[0].set_ylabel('λ(t) — intensity')
axes[0].set_title(
    f"Hawkes Process Intensity (first 60s)\n"
    f"μ={params['mu']:.3f}  α={params['alpha']:.3f}  "
    f"β={params['beta']:.3f}  n=α/β={params['branching_ratio']:.3f}"
)
axes[1].eventplot(sample_ts, lineoffsets=0.5, linewidths=0.5, colors=['#00ff88'])
axes[1].set_ylabel('Trades'); axes[1].set_xlabel('Time (seconds)')
axes[1].set_yticks([])
plt.tight_layout()
plt.savefig('../data/processed/04_hawkes_intensity.png', dpi=150, bbox_inches='tight')
plt.show()
```
